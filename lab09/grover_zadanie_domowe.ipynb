```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(2, name="Grover n=2 (|10>)")

# Superpozycji
qc.h([0, 1])
qc.barrier()

# Wyrocznia dla stanu |10>
qc.x(0)
qc.cz(0, 1)
qc.x(0)
qc.barrier()

# Inwersja
qc.h([0, 1])
qc.x([0, 1])
qc.cz(0, 1)
qc.x([0, 1])
qc.h([0, 1])

display(qc.draw(output="mpl"))
```

```python
from qiskit.quantum_info import Statevector

state = Statevector.from_instruction(qc)

display(state.draw("latex"))
```
