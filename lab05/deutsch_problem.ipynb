```python
from qiskit import QuantumCircuit, QuantumRegister

# "f_0" and "f_3" are constant functions
# "f_1" and "f_2" are balanced functions
U_f = "f_1"

qc = QuantumCircuit(QuantumRegister(2, "q"))

# Initialize qubit q_1 to |1>
qc.x(1)

# Apply Hadamard gates
qc.h(0)
qc.h(1)

qc.barrier()

# Choose U_f gate
if U_f == "f_0":
    pass
elif U_f == "f_1":
    qc.cx(0, 1)
elif U_f == "f_2":
    qc.x(1)
    qc.cx(0, 1)
elif U_f == "f_3":
    qc.x(1)

qc.barrier()

# Apply Hadamard gates again
qc.h(0)
qc.h(1)

display(qc.draw(output="mpl"))
```

```python
from qiskit.quantum_info import Statevector

state = Statevector.from_instruction(qc)

ideal_dist = state.probabilities_dict()

# |q_1 q_0>
# For constants q_0 is always 0
# For balanced q_0 is always 1
display(state.draw("latex"))
```
